##**Vanishing Gradient Problem in ANN/CNN/RNN:**
- When gradient(loss with respose to features) is very low in this case then learning process becaomes very slow.(Vanishing gradient)
- When gradient is very high in that case learning process becomes too high(it is called exploding gradient.)
- In RNN, if sentence is too long then significance of first word will be low at the end word of the same sentence.
- RNN has short memory.not able to save long memory.
- Techniques to handle short term memory problem in RNN.
  - LSTM
  - GRU

**LSTM(Long short term memory):**
  - Create a new state to store kry points in long term memory.
  - Forget Gate: sigmoid (for removing previous keypoints)
  - Input Gate: sigmoid + tanh (for adding current keypoints)
  - Output Gate: sigmoid + tanh (prepare final output)

**GRU(Gated Recurrent Units):**
  - it is light weight LSTM.combied long&short term memory.
  - only one channel for both long and short term memory.
  - Reset Gate: sigmoid (reset keypoints)
  - update Gate: sigmoid (updated keypoints)
  - hadamard matrix product: element by element product
  - https://en.wikipedia.org/wiki/Hadamard_product_(matrices)
