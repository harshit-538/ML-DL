# **Activation Functions**

Mostly used to make any variable in range of 0 to 1 or -1 to 1.it used to activate a neuron

### **1. Sigmoid**



```python
import math
def sigmoid(x):
  return 1 / (1 + math.exp(-x))
```

```python
sigmoid(-82)
```

### **2.  Tanh**

```python
import math
def tanh(x):
    return (math.exp(x) - math.exp(-x))/(math.exp(x) + math.exp(-x))
```

```python
tanh(-0.2)
```

### **3. ReLU & Leaky ReLU**

```python
def relu(x):
    return max(0,x)

def LeakyReLU(x):
    return max(0.1*x,x)
```

```python
relu(543)
LeakyReLU(543)
```

# **Loss/Cost Functions**

```python
import numpy as np

y_pred = np.array([1,1,0,0,1])
y_true = np.array([0.3,0.7,1,0,0.5])
```

### **Mean Absolute Error(MAE)**

```python
def mae(y_pred,y_true):
  total_error = 0
  for yp,yt in zip(y_pred,y_true):
    total_error += abs(yp - yt)

  mae = total_error / len(y_pred)
  return mae
```

```python
mae(y_pred,y_true)
```

```python
np.mean(np.abs(y_pred-y_true))
```

### **Mean Squared Error(MSE)**

```python
def mse(y_pred,y_true):
  total_error = 0
  for yp,yt in zip(y_pred,y_true):
    total_error += (yp - yt)**2

  mse = total_error / len(y_pred)
  return mse
```

```python
mse(y_pred,y_true)
```

```python
np.mean(np.square(y_pred-y_true))
```

### **Log Loss Function/Binary cross Entropy**

```python
epsilon = 1e-15
y_pred_new = [max(i,epsilon) for i in y_pred]
y_pred_new = [min(i,1-epsilon) for i in y_pred_new]
y_pred_new = np.array(y_pred_new)

log_loss = -np.mean(y_true*np.log(y_pred_new) + (1-y_true)*np.log(1-y_pred_new))
log_loss
```
